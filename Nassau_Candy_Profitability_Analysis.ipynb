**Objective:**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
df = pd.read_csv("Nassau Candy Distributor.csv")

print("Dataset loaded successfully!")
print(df.shape)
display(df.head())
```

```python
print("Unique Divisions:")
print(df['Division'].unique())

print("\nNumber of Divisions:")
print(df['Division'].nunique())

print("\nNumber of Unique Products:")
print(df['Product Name'].nunique())
```

```python
# Check missing values in each column
missing_values = df.isnull().sum()

print("Missing Values:")
print(missing_values)

# Calculate total missing values
print("\nTotal Missing Values:", df.isnull().sum().sum())
```

```python
# Check duplicate records
duplicate_count = df.duplicated().sum()

print("Total Duplicate Records:", duplicate_count)
```

```python
# Convert date columns into datetime format
df['Order Date'] = pd.to_datetime(
    df['Order Date'], format='%d-%m-%Y'
)

df['Ship Date'] = pd.to_datetime(
    df['Ship Date'], format='%d-%m-%Y'
)

# Check updated data types
print(df[['Order Date', 'Ship Date']].dtypes)

# Display first five dates
df[['Order Date', 'Ship Date']].head()
```

```python
# Check records where shipping date is before order date
invalid_dates = df[df['Ship Date'] < df['Order Date']]

print("Invalid Date Records:", len(invalid_dates))

# Calculate shipping duration
df['Shipping Days'] = (
    df['Ship Date'] - df['Order Date']
).dt.days

print("\nShipping Duration Summary:")
print(df['Shipping Days'].describe())

# Display first five records
df[['Order Date', 'Ship Date', 'Shipping Days']].head()
```

```python
# Check zero and negative values
columns = ['Sales', 'Units', 'Gross Profit', 'Cost']

for col in columns:
    print(f"\n{col}")
    print("Zero values:", (df[col] == 0).sum())
    print("Negative values:", (df[col] < 0).sum())
```

```python
# Calculate gross profit independently
df['Calculated Profit'] = df['Sales'] - df['Cost']

# Calculate the difference
df['Profit Difference'] = (
    df['Gross Profit'] - df['Calculated Profit']
).round(2)

# Count mismatched records
mismatch_count = (df['Profit Difference'] != 0).sum()

print("Profit Calculation Mismatches:", mismatch_count)

# Display mismatched records, if any
print(df[df['Profit Difference'] != 0][
    ['Sales', 'Cost', 'Gross Profit', 'Calculated Profit']
].head())
```

```python
# Gross Margin Percentage
df['Gross Margin %'] = (
    df['Gross Profit'] / df['Sales']
) * 100

# Profit Per Unit
df['Profit Per Unit'] = (
    df['Gross Profit'] / df['Units']
)

# Revenue Contribution
df['Revenue Contribution %'] = (
    df['Sales'] / df['Sales'].sum()
) * 100

# Profit Contribution
df['Profit Contribution %'] = (
    df['Gross Profit'] / df['Gross Profit'].sum()
) * 100

# Display new profitability metrics
df[[
    'Product Name',
    'Sales',
    'Units',
    'Gross Profit',
    'Cost',
    'Gross Margin %',
    'Profit Per Unit',
    'Revenue Contribution %',
    'Profit Contribution %'
]].head()
```

```python
# Overall business performance

total_sales = df['Sales'].sum()
total_cost = df['Cost'].sum()
total_profit = df['Gross Profit'].sum()

overall_margin = (total_profit / total_sales) * 100

print("Total Sales:", round(total_sales, 2))
print("Total Cost:", round(total_cost, 2))
print("Total Gross Profit:", round(total_profit, 2))
print("Overall Gross Margin:", round(overall_margin, 2), "%")
```

```python
# Group data by product name
product_analysis = df.groupby('Product Name').agg({
    'Sales': 'sum',
    'Units': 'sum',
    'Cost': 'sum',
    'Gross Profit': 'sum'
}).reset_index()

# Calculate product-level gross margin
product_analysis['Gross Margin %'] = (
    product_analysis['Gross Profit'] /
    product_analysis['Sales']
) * 100

# Calculate profit per unit
product_analysis['Profit Per Unit'] = (
    product_analysis['Gross Profit'] /
    product_analysis['Units']
)

# Sort products by gross profit
product_analysis = product_analysis.sort_values(
    by='Gross Profit', ascending=False
)

# Display results
product_analysis
```

```python
plt.figure(figsize=(12, 7))

sns.barplot(
    data=product_analysis,
    x='Gross Profit',
    y='Product Name'
)

plt.title('Product-wise Gross Profit Analysis')
plt.xlabel('Total Gross Profit')
plt.ylabel('Product Name')
plt.tight_layout()
plt.show()
```

```python
# Group data by division
division_analysis = df.groupby('Division').agg({
    'Sales': 'sum',
    'Units': 'sum',
    'Cost': 'sum',
    'Gross Profit': 'sum'
}).reset_index()

# Calculate gross margin
division_analysis['Gross Margin %'] = (
    division_analysis['Gross Profit'] /
    division_analysis['Sales']
) * 100

# Calculate profit per unit
division_analysis['Profit Per Unit'] = (
    division_analysis['Gross Profit'] /
    division_analysis['Units']
)

# Sort by gross profit
division_analysis = division_analysis.sort_values(
    by='Gross Profit', ascending=False
)

division_analysis
```

```python
plt.figure(figsize=(9, 5))

sns.barplot(
    data=division_analysis,
    x='Division',
    y='Gross Profit'
)

plt.title('Division-wise Gross Profit Analysis')
plt.xlabel('Division')
plt.ylabel('Total Gross Profit')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.barplot(
    data=division_analysis,
    x='Division',
    y='Gross Margin %'
)

plt.axhline(
    y=overall_margin,
    linestyle='--',
    label='Overall Gross Margin'
)

plt.title('Division-wise Gross Margin Comparison')
plt.xlabel('Division')
plt.ylabel('Gross Margin (%)')
plt.legend()

plt.tight_layout()
plt.show()
```

```python
# Sort products by gross profit
pareto_analysis = product_analysis.sort_values(
    by='Gross Profit',
    ascending=False
).copy()

# Calculate profit contribution percentage
pareto_analysis['Profit Contribution %'] = (
    pareto_analysis['Gross Profit'] /
    pareto_analysis['Gross Profit'].sum()
) * 100

# Calculate cumulative profit contribution
pareto_analysis['Cumulative Profit %'] = (
    pareto_analysis['Profit Contribution %'].cumsum()
)

# Display Pareto analysis
pareto_analysis[[
    'Product Name',
    'Gross Profit',
    'Profit Contribution %',
    'Cumulative Profit %'
]]
```

```python
fig, ax1 = plt.subplots(figsize=(14, 7))

# Bar chart: Gross Profit
ax1.bar(
    pareto_analysis['Product Name'],
    pareto_analysis['Gross Profit']
)

ax1.set_xlabel('Product Name')
ax1.set_ylabel('Gross Profit')
ax1.set_title('Pareto Analysis - Product Profit Concentration')

plt.xticks(rotation=75, ha='right')

# Second axis: Cumulative Profit %
ax2 = ax1.twinx()

ax2.plot(
    pareto_analysis['Product Name'],
    pareto_analysis['Cumulative Profit %'],
    marker='o',
    linestyle='-'
)

ax2.set_ylabel('Cumulative Profit Contribution (%)')
ax2.axhline(y=80, linestyle='--', label='80% Reference Line')
ax2.legend(loc='center right')

plt.tight_layout()
plt.show()
```

```python
# Calculate cost percentage for each product
product_analysis['Cost %'] = (
    product_analysis['Cost'] /
    product_analysis['Sales']
) * 100

# Display cost and profitability metrics
cost_analysis = product_analysis[[
    'Product Name',
    'Sales',
    'Cost',
    'Gross Profit',
    'Cost %',
    'Gross Margin %'
]].sort_values(by='Cost %', ascending=False)

cost_analysis
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=product_analysis,
    x='Cost %',
    y='Gross Margin %',
    hue='Division' if 'Division' in product_analysis.columns else None,
    s=120
)

plt.title('Product Cost Percentage vs Gross Margin')
plt.xlabel('Cost Percentage (%)')
plt.ylabel('Gross Margin (%)')

plt.tight_layout()
plt.show()
```

```python
# Calculate gross margin for each transaction
df['Transaction Margin %'] = (
    df['Gross Profit'] / df['Sales']
) * 100

# Calculate margin volatility for each product
margin_volatility = df.groupby('Product Name').agg({
    'Transaction Margin %': ['mean', 'std', 'min', 'max']
}).reset_index()

# Rename columns
margin_volatility.columns = [
    'Product Name',
    'Average Margin %',
    'Margin Volatility',
    'Minimum Margin %',
    'Maximum Margin %'
]

# Sort by margin volatility
margin_volatility = margin_volatility.sort_values(
    by='Margin Volatility',
    ascending=False
)

margin_volatility
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    data=margin_volatility,
    x='Margin Volatility',
    y='Product Name'
)

plt.title('Product-wise Margin Volatility Analysis')
plt.xlabel('Margin Volatility (Standard Deviation)')
plt.ylabel('Product Name')

plt.tight_layout()
plt.show()
```

```python
# Calculate revenue and profit contribution by division

division_analysis['Revenue Contribution %'] = (
    division_analysis['Sales'] / total_sales
) * 100

division_analysis['Profit Contribution %'] = (
    division_analysis['Gross Profit'] / total_profit
) * 100

division_analysis
```

```python
# Create overall business KPI summary

executive_summary = pd.DataFrame({
    'KPI': [
        'Total Sales',
        'Total Cost',
        'Total Gross Profit',
        'Overall Gross Margin',
        'Total Transactions',
        'Unique Products',
        'Number of Divisions',
        'Top Profit Generating Product',
        'Highest Margin Product',
        'Lowest Margin Product'
    ],

    'Value': [
        round(total_sales, 2),
        round(total_cost, 2),
        round(total_profit, 2),
        round(overall_margin, 2),
        len(df),
        df['Product Name'].nunique(),
        df['Division'].nunique(),
        product_analysis.iloc[0]['Product Name'],
        product_analysis.loc[
            product_analysis['Gross Margin %'].idxmax(),
            'Product Name'
        ],
        product_analysis.loc[
            product_analysis['Gross Margin %'].idxmin(),
            'Product Name'
        ]
    ]
})

executive_summary
```

## Conclusion

The Product Line Profitability and Margin Performance Analysis of Nassau Candy Distributor was conducted using Python, Pandas, NumPy, Matplotlib and Seaborn.

The dataset contained 10,194 transactions, 15 unique products and 3 divisions. The analysis identified total sales of 141,783.63, total gross profit of 93,442.80 and an overall gross margin of 65.91%.

The Chocolate division was the strongest contributor, generating 95.06% of total gross profit. Wonka Bar - Scrumdiddlyumptious was the highest gross-profit-generating product, while Everlasting Gobstopper recorded the highest gross margin of 80%.

Kazookles had the lowest gross margin of 7.69%, indicating a need for further cost and pricing investigation. Pareto analysis also showed that the top five products contributed approximately 95.06% of total gross profit.

Overall, the analysis demonstrates the importance of evaluating product profitability, cost structures, gross margins and profit concentration rather than relying only on sales volume. The findings can support better product-level and division-level business decisions.
